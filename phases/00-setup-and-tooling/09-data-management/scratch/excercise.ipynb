```python
#Load the glue dataset with the mrpc config and inspect the first 5 examples
from datasets import load_dataset

dataset = load_dataset("nyu-mll/glue", "mrpc")
for i in range(5):
    print(dataset["train"][i])
```

```python
#Stream the c4 dataset and count how many examples you can process in 10 seconds
import time

en = load_dataset("allenai/c4", "en",split="train", streaming=True)
count = 0
start_time = time.time()
for i, example in enumerate(en):
    count+=1
    if time.time() - start_time > 10:
        break
print(f"Processed {count} examples in 10 seconds")
```

```python
dataset = dataset["train"]
dataset.to_csv("imdb_train.csv")
dataset.to_json("imdb_train.json")
dataset.to_parquet("imdb_train.parquet")
```

```python
#Convert a dataset to Parquet and compare the file size to CSV
from pathlib import Path

csv_size = Path("imdb_train.csv").stat().st_size
parquet_size = Path("imdb_train.parquet").stat().st_size
print(f"CSV size: {csv_size} bytes")
print(f"Parquet size: {parquet_size} bytes")
print(f"Parquet is {csv_size/parquet_size:.2f} times smaller than CSV")
```

```python
#Create a 70/15/15 train/val/test split with a fixed seed and verify the sizes
SEED = 42
split = dataset.train_test_split(test_size=0.3, seed=SEED)
train_val_split = split["train"].train_test_split(test_size=0.15, seed=SEED)
train_dataset = train_val_split["train"]
val_dataset = train_val_split["test"]
test_dataset = split["test"]
print(f"Train size: {len(train_dataset)}")
print(f"Validation size: {len(val_dataset)}")
print(f"Test size: {len(test_dataset)}")
```
